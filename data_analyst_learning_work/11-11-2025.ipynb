```python
# What is CRUD?
# C → Create
# R → Read
# U → Update
# D → Delete
#  1. READ  (Fully supported)

#  Both loc and iloc are mainly used for reading data

# df.loc[0:2, 'Name']
# df.iloc[0:3, 0:2]

#  Most common use

#  2. UPDATE  (Supported)

#  You can modify values

# df.loc[0, 'Age'] = 25
# df.iloc[0, 1] = 25

#  Works perfectly

#  3. CREATE  (Limited)

#  You can add new rows/columns using loc

# df.loc[5] = ['Ajay', 22]

#  loc can create
#  (!) iloc cannot create new rows

#  4. DELETE  (Not directly)

#  loc / iloc cannot delete directly

         # or 

# peration	        loc 	              iloc
# Read	           ✅     	             ✅
# Update	      ✅  	                ✅
# Create	     ✅ (limited)	       ❌
# Delete	    ❌	                  ❌
```

```python
# Missing ValueDetected by 
#                           isna()    ?Detected by isnull()?
# NaN                         Yes                 Yes
# None                        Yes                  Yes
# NaT (datetime)              Yes                  Yes
# pd.NA                       Yes                  Yes

# Historical reasons + user convenience.
# Here's the story:

# isnull() came first — it was the original function when pandas was created.
# isna() was added later — pandas wanted to make it more intuitive, because NA is a more standard term for "missing value" in data science (especially for people coming from R language, where is.na() is used).
# They couldn't remove isnull() — because millions of people were already using it in their code. Removing it would break existing code, so they kept both.
```
