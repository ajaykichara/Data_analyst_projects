# bivariate analysis

```python
# 1.scatterplot(numerical_columns - numeric_columns)
sns.scatterplot(x=tips['total_bill'],y=tips['tip']) #Tip depends on bill ,“Bill ke respect mein tip nikalti hai”
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', size='size')
# when we use   data=df  then  we  not  wrte  dataframe  name  for column like hue=df['tip']
# if  not  use data =df  then  write  like  hue  =  df['tip']
```

```python
# Concept: Univariate, Bivariate, and Multivariate analysis depend on the number of UNIQUE variables (columns),
# not on the number of parameters used in the visualization.

# - Univariate   → 1 unique variable
# - Bivariate    → 2 unique variables
# - Multivariate → 3 or more unique variables

# Important:
# Repeating the same column in multiple parameters (x, y, hue, size, etc.)
# does NOT increase the number of variables.

# Example:
# x='total_bill', y='tip', hue='total_bill' → BIVARIATE (2 unique variables)
# x='total_bill', y='tip', hue='sex' → MULTIVARIATE (3 unique variables)
```

```python
# # hue → color (category)
# # size → magnitude (numeric)
# # style → shape (category)

# Numeric + Numeric → scatterplot
# Category present → use hue/style
# Quantity → use size

# how many variable  in plot(function  is  calle  vaiabte  if 2  = bi  ,if 1  = uni  ,  3or 4 then  multivaiate)
# sns.scatterplot(data=df, x='age', y='salary')
# data=df → source ❌
# age → variable ✅
# salary → variable ✅

# 👉 Total variables = 2 → Bivariate
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex',style = 'smoker',size='size') #Multivariate
```

```python
# blue circle  =  male + smoking ,blue cross  =  male+withoutsmoking 
#  1. Univariate

#   1 variable analysis

# Focus: only one column
# Goal: understand distribution

#  Example:

# Age distribution
# Salary distribution
#  2. Bivariate

#  2 variables analysis

# Focus: relationship between two columns
# Goal: find correlation / comparison

#  Example:

# Age vs Salary
# Bill vs Tip
#  3. Multivariate

#  3 or more variables

# Focus: multiple relationships
# Goal: deep insights

#  Example:

# Age, Salary, Experience together
```

```python
# boxplot(numeric-categoric)
sns.boxplot(data= tips,x='time',y='tip',hue='day')
```

```python
# sns.displot(numerical - categorical)
sns.displot(tips['sex']) # Male ≈ 150+ and Female ≈ 90 in dataset
```

```python
sns.displot(tips[tips['total_bill'] > 40]['tip'], kind='kde') #new  
```

```python
#crosstab(catergorical -  categoical)
p  = pd.crosstab(tips['day'],tips['time'])
p = pd.crosstab(tips['day'], [tips['time'], tips['sex']])
p
```

```python
sns.heatmap(p)
```
