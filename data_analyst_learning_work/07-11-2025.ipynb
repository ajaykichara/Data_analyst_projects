```python
df2 = pd.read_csv("C:\\Users\\Admin\\pandas\\shoes_2000_to_5000.csv")
```

```python
df2 = df2.rename(columns=lambda col: (
    col.replace("CIaYa1 href", "product_link")
       .replace("MZeksS src", "image_url")
       .replace("Fo1I0b", "brand")
       .replace("atJtCj href", "product_page")
       .replace("atJtCj", "product_name")
       .replace("Q3Sg7S src", "extra_image")
       .replace("hZ3P6w", "price")
       .replace("kRYCnD", "original_price")
       .replace("HQe8jr", "discount")
       .replace("HZ0E6r", "rating")
       .replace("yZLfgx", "size_options")
       .replace("g_3iVf", "size_label")
))
```

```python
print(df2.columns.to_list())
```

```python
df2.head(2)
```

```python
df2 = df2.drop(columns=[col for col in df2.columns if 'link' in col or 'image' in col or 'page' in col])
```

```python
import pandas as pd

# Step 1: Convert wide → long
final_data = []

for i in range(len(df2)):
    row = df2.iloc[i]
    
    for j in ['', ' 2', ' 3', ' 4']:
        try:
            name = row[f'product_name{j}']
            brand = row[f'brand{j}']
            price = row[f'price{j}']
            original_price = row.get(f'original_price{j}', None)
            discount = row.get(f'discount{j}', None)
            
            if pd.notna(name):
                final_data.append([name, brand, price, original_price, discount])
        
        except:
            continue

# Step 2: Create DataFrame
final_df = pd.DataFrame(final_data, columns=[
    'product_name', 'brand', 'price', 'original_price', 'discount'
])

# Step 3: Create numeric columns (keep original also)
final_df['price_num'] = final_df['price'].str.replace('₹', '').str.replace(',', '').astype(float)

final_df['original_price_num'] = final_df['original_price'].str.replace('₹', '').str.replace(',', '').astype(float)

final_df['discount_num'] = final_df['discount'].str.replace('% off', '').astype(float)
final_df = final_df.drop(columns=['original_price_num', 'discount_num','price_num'])

final_df['original_price'] = final_df['original_price'].fillna("Not Available")
final_df['discount'] = final_df['discount'].fillna("No Discount")

final_df = final_df.drop_duplicates()
# Step 4: Output
final_df
```
